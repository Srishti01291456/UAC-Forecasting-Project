```python
df['lag1'] = df['Children in HHS Care'].shift(1)

df['lag7'] = df['Children in HHS Care'].shift(7)

df['lag14'] = df['Children in HHS Care'].shift(14)
```

```python
df['roll7'] = (
    df['Children in HHS Care']
      .rolling(7)
      .mean()
)

df['roll14'] = (
    df['Children in HHS Care']
      .rolling(14)
      .mean()
)
```
