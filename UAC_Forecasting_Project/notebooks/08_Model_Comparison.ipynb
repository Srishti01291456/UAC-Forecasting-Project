```python
results = {
    "Model": [
        "Naive",
        "ARIMA",
        "SARIMA",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        150,
        120,
        110,
        85,
        90
    ],
    "RMSE": [
        180,
        140,
        125,
        95,
        100
    ],
    "MAPE": [
        12.5,
        9.8,
        8.7,
        6.1,
        6.5
    ]
}
```

```python
import pandas as pd

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
comparison_df.sort_values(
    by="RMSE"
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    comparison_df["Model"],
    comparison_df["RMSE"]
)

plt.title("RMSE Comparison")
plt.ylabel("RMSE")
plt.xticks(rotation=20)

plt.show()
```
